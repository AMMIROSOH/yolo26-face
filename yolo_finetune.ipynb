```python
from ultralytics import YOLO
```

```python
DEVICE = "cpu" # cuda:0 
model = YOLO("yolov8n-pose.pt")
```

```python
# Fine-tune on your dataset
model.train(
    data='D:/Datasets/yolo26/wider/dataset/data.yml',
    epochs=50,
    imgsz=640,
    batch=16,
    lr0=0.0005,
    workers=8, 
    # cache="disk",
    pretrained=True,
    project='face',
    name='finetune',
    device=DEVICE,
)
```
